```python
import pandas as pd
import numpy as np
from utils import save_npz
```

Normalize the data for faster and more stable training

```python
train = np.load("train.npz") 
X_train = train["X"] 
y_train = train["y"]

val = np.load("val.npz") 
X_val = val["X"] 
y_val = val["y"]

test = np.load("test.npz") 
X_test = test["X"] 
y_test = test["y"]
```

```python
def data_preprocessing(x_train, x_val, x_test):
    mean = np.mean(x_train, axis=0)
    std = np.std(x_train, axis=0)
    x_train_processed = (x_train - mean) / std
    # we subtract the mean of the training data to prevent the model from accessing information from the test data, ensuring a precise evaluation.
    x_val_processed = (x_val - mean) / std 
    x_test_processed = (x_test - mean) / std
    print("Data preprocessing completed.")
    return (x_train_processed, x_val_processed, x_test_processed)

X_train_processed, X_val_processed, X_test_processed = data_preprocessing(X_train, X_val, X_test)
```

```python
save_npz("train_processed.npz", X_train_processed, y_train)
save_npz("val_processed.npz", X_val_processed, y_val)
save_npz("test_processed.npz", X_test_processed, y_test)
```
