```python
import pandas as pd
import numpy as np
from utils import load_data
from sklearn.model_selection import train_test_split
from utils import save_npz
```

Stratified split for train and validation

```python
x_train, y_train=load_data("data", "data/Train.csv", 30, 30)
x_test, y_test=load_data("data", "data/Test.csv", 30, 30)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    x_train, y_train,
    test_size=0.125,
    stratify=y_train,
    random_state=42
)
```

Store the split data into .npz file, for reusability

```python
X_train = np.array(X_train)
X_val = np.array(X_val)
x_test = np.array(x_test)

y_train = np.array(y_train)
y_val = np.array(y_val)
y_test = np.array(y_test)

save_npz("train.npz", X_train, y_train)
save_npz("val.npz", X_val, y_val)
save_npz("test.npz", x_test, y_test)

print("output finished!")
```
